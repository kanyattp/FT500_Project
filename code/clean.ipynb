```python
# Verification of re-indexing in python if needed
import pandas as pd

excel_path = 'fortune500.xlsx'
df = pd.read_excel(r"d:\M\python\fortune500 (1).xlsx")
df_latest = df[df['Year'] == df['Year'].max()].copy()

df_latest['Revenue (in millions)'] = pd.to_numeric(df_latest['Revenue (in millions)'], errors='coerce')
df_latest['Profit (in millions)'] = pd.to_numeric(df_latest['Profit (in millions)'], errors='coerce')

def assign_industry(company):
    c = str(company).lower()
    if any(k in c for k in ['exxon', 'chevron', 'conoco', 'valero', 'oil']):
        return 'พลังงานและปิโตรเลียม'
    elif any(k in c for k in ['general electric', 'boeing', 'caterpillar']):
        return 'อุตสาหกรรมและการผลิต'
    elif any(k in c for k in ['general motors', 'ford', 'chrysler']):
        return 'ยานยนต์'
    elif any(k in c for k in ['mckesson', 'cardinal health', 'pfizer', 'johnson & johnson']):
        return 'เวชภัณฑ์และยา'
    elif any(k in c for k in ['dow chemical', 'du pont', 'dupont', 'ppg', 'monsanto']):
        return 'เคมีภัณฑ์'
    return None

df_latest['Industry'] = df_latest['Company'].apply(assign_industry)
df_filtered = df_latest.dropna(subset=['Industry']).copy()

df_filtered['Rev_Rank'] = df_filtered.groupby('Industry')['Revenue (in millions)'].rank(ascending=False)
df_filtered['Prof_Rank'] = df_filtered.groupby('Industry')['Profit (in millions)'].rank(ascending=False)
df_filtered['Combined_Score'] = df_filtered['Rev_Rank'] + df_filtered['Prof_Rank']

top1_overall = (
    df_filtered
    .sort_values(by=['Industry', 'Combined_Score', 'Revenue (in millions)'], ascending=[True, True, False])
    .groupby('Industry')
    .first()
    .reset_index()
)

# Reset index to 1-5
top1_overall.index = range(1, len(top1_overall) + 1)
print(top1_overall[['Industry', 'Company', 'Revenue (in millions)', 'Profit (in millions)', 'Combined_Score']])
```

```python
import pandas as pd
import numpy as np

# 1. โหลดข้อมูลจากไฟล์ Excel
file_path = r'd:\M\python\fortune500 (1).xlsx'
df = pd.read_excel(file_path)

# 2. ตรวจสอบสภาพข้อมูลเบื้องต้น
print("--- ข้อมูลเบื้องต้นก่อนทำความสะอาด ---")
print(df.info())
print("\nจำนวนแถวที่ซ้ำกัน:", df.duplicated().sum())
print("จำนวนค่าว่าง (Missing Values):")
print(df.isnull().sum())

# 3. จัดประเภทข้อมูลและแปลงคอลัมน์ตัวเลข (Data Type Conversion)
# errors='coerce' จะช่วยเปลี่ยนค่าที่ไม่ใช่ตัวเลขให้เป็น NaN เพื่อป้องกัน Error
df['Revenue (in millions)'] = pd.to_numeric(df['Revenue (in millions)'], errors='coerce')
df['Profit (in millions)'] = pd.to_numeric(df['Profit (in millions)'], errors='coerce')
df['Year'] = df['Year'].astype(int)
df['Rank'] = df['Rank'].astype(int)

# 4. จัดการและล้างข้อความ (String Cleaning)
# ตัดช่องว่างหน้า-หลังข้อความออก
df['Company'] = df['Company'].astype(str).str.strip()

# 5. ลบแถวที่ซ้ำกัน (Remove Duplicates)
df = df.drop_duplicates()

# 6. เติมค่าว่าง (Handle Missing Values)
# หากมีค่าว่างในคอลัมน์กำไร ให้แทนที่ด้วยค่า Median หรือ 0
if df['Profit (in millions)'].isnull().sum() > 0:
    df['Profit (in millions)'] = df['Profit (in millions)'].fillna(df['Profit (in millions)'].median())

# 7. กรองข้อมูลที่ผิดปกติ (Filter Outliers/Invalid Ranges)
df = df[(df['Year'] >= 1955) & (df['Year'] <= 2009)]

# 8. บันทึกข้อมูลที่ทำความสะอาดแล้วเป็นไฟล์ใหม่
output_path = 'fortune500_cleaned.xlsx'
df.to_excel(output_path, index=False)
```

```python
import pandas as pd

# 1. อ่านไฟล์ข้อมูล Excel (เลือกใช้ไฟล์ที่คุณทำความสะอาดแล้วหรือไฟล์ที่มี 5 บริษัท)
file_path = r'd:\python1\data\clean\No.1_Python.xlsx'
df = pd.read_excel(file_path)

# 2. แปลงข้อมูลเชิงตัวเลขให้ถูกต้อง
df['Revenue (in millions)'] = pd.to_numeric(df['Revenue (in millions)'], errors='coerce')
df['Profit (in millions)'] = pd.to_numeric(df['Profit (in millions)'], errors='coerce')

# 3. คำนวณ mean, max, median, min ของ Revenue และ Profit แยกตามบริษัท
company_stats = df.groupby('Company').agg({
    'Revenue (in millions)': ['mean', 'max', 'median', 'min'],
    'Profit (in millions)': ['mean', 'max', 'median', 'min']
}).round(2)  # ปัดเศษทศนิยม 2 ตำแหน่ง

# 4. แสดงผลลัพธ์
print(company_stats)

# 5. (Optional) บันทึกผลสรุปสถิติออกเป็นไฟล์ Excel ใหม่
company_stats.to_excel('company_summary_stats.xlsx')
```

```python
import pandas as pd

# 1. อ่านไฟล์ข้อมูล Excel
file_path = r'd:\python1\data\clean\No.1_Python.xlsx'
df = pd.read_excel(file_path)

# 2. แปลงข้อมูลเชิงตัวเลขให้ถูกต้อง
df['Revenue (in millions)'] = pd.to_numeric(df['Revenue (in millions)'], errors='coerce')
df['Profit (in millions)'] = pd.to_numeric(df['Profit (in millions)'], errors='coerce')

# 3. คำนวณ mean, std, max, median, min ของ Revenue และ Profit
company_stats = df.groupby('Company').agg({
    'Revenue (in millions)': ['mean', 'std', 'median', 'max', 'min'],
    'Profit (in millions)': ['mean', 'std', 'median', 'max', 'min']
}).round(2)

company_stats = company_stats.rename(columns={'std': 'SD'})

# 4. แสดงผลลัพธ์
print(company_stats)

# 5. บันทึกผลสรุปสถิติออกเป็นไฟล์ Excel
company_stats.to_excel('company_summary_stats_with_sd.xlsx')
```
